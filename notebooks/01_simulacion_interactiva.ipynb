# Simulación interactiva de tenis de mesa

Ajusta los controles y observa cómo cambian la trayectoria, la velocidad y el giro. El notebook usa el paquete de `src/`; no duplica la física.

```python
from pathlib import Path
import sys
import numpy as np
import matplotlib.pyplot as plt
from ipywidgets import FloatSlider, interact

root = Path.cwd().resolve().parent if Path.cwd().name == 'notebooks' else Path.cwd().resolve()
sys.path.insert(0, str(root / 'src'))

from table_tennis_sim import SimulationParameters, run_simulation
from table_tennis_sim.visualization import plot_trajectory_3d, plot_time_series
```

```python
@interact(
    vx=FloatSlider(value=7000, min=1000, max=10000, step=250, description='vx [mm/s]'),
    vy=FloatSlider(value=-3000, min=-6000, max=6000, step=250, description='vy [mm/s]'),
    vz=FloatSlider(value=-3000, min=-6000, max=2000, step=250, description='vz [mm/s]'),
    omega_z=FloatSlider(value=75, min=-150, max=150, step=5, description='omega z [rev/s]'),
    drag=FloatSlider(value=2.7, min=0, max=6, step=0.1, description='arrastre'),
    magnus=FloatSlider(value=0.01, min=0, max=0.05, step=0.005, description='Magnus'),
    table_restitution=FloatSlider(value=0.77, min=0, max=1, step=0.01, description='restit. mesa'),
    net_restitution=FloatSlider(value=0.5, min=0, max=1, step=0.01, description='restit. red'),
    table_friction=FloatSlider(value=0.25, min=0, max=1, step=0.01, description='fricción mesa'),
    total_time=FloatSlider(value=1.5, min=0.2, max=3, step=0.1, description='tiempo [s]'),
    dt=FloatSlider(value=0.005, min=0.001, max=0.02, step=0.001, description='paso [s]'),
)
def explorar(vx, vy, vz, omega_z, drag, magnus, table_restitution, net_restitution, table_friction, total_time, dt):
    params = SimulationParameters(
        initial_velocity=np.array([vx, vy, vz]),
        initial_omega=np.array([0.0, 0.0, omega_z]) * 2 * np.pi,
        drag=drag, magnus=magnus, table_restitution=table_restitution,
        net_restitution=net_restitution, table_friction=table_friction,
        total_time=total_time, dt=dt,
    )
    result = run_simulation(params)
    print(f'Parámetros: vx={vx:.0f}, vy={vy:.0f}, vz={vz:.0f} mm/s; giro z={omega_z:.0f} rev/s; Magnus={magnus:.3f}.')
    print('Al aumentar Magnus o el giro cambia la curvatura; al aumentar arrastre disminuye el alcance.')
    plot_trajectory_3d(result, params)
    plot_time_series(result)
    plt.show()
```
